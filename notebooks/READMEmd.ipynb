#Netflix Movie and TV Show Analysis


##Layihə haqqında
Bu layihədə Netflix platformasında olan dataları analiz edilib.Ilk növbədə metodlar vasitəsi ilə data haqqında məlumat,Null dəyərləri,vizuallaşdırma işləri,ən çox film çəkmiş TOP 10 ölkə, aylıq və illik yayımlanan filmlər araşdırılmışdır.
Layihənin əsas məqsədi yayımlanan Film və TV Şoular haqqında məlumat almaq və data əsasında düzgün qərarların verilməsidir.

---

##Layihənin məqsədləri
Analiz zaman aşağıdakı suallara cavab axtarılmışdır:

1.Datanın neçə tipi var və vizual olaraq paylanması necədir?

2.Null(çatışmayan) dəyərlərin qarşısı necə alınır?

3.Ən çox hansı ölkənin filmləri yayımlanır?

4.Hər il və hər ay yayımlanan Film sayı nə qədərdir?

5.`Pivot Table` və `Heatmap` vasitəsilə mövsumi trendləri necə göstərmək olar?

---

#Dataset
Layihədə Kaggle vebsite-da olan Netflix Movie and TV Show dataset-i istifadə edilmişdir.

Dataset aşağıdakı əsas sütunlardan ibarətdir:


`show_id` ID sütunu

`type`  Növü(Movie or TV Show)

`title` Film və ya şou adı

`director`  Müəllifi

`cast` Rol alan aktyorlar

`country` Ölkə

`date_added` Netflix-ə əlavə edildiyi tarix

`release_year` İlk yayımlandığı il

`rating` Yaş raytinqi

`duration` Film müddəti və ya serialın sezonu

`listed_in` Janr/Kategoriya

`description` Qısa təsviri

#Data Analitika Prosesi

1.Data Cleaning
*   Null dəyərlər təmizləndi
*   `data_added` sütunu `datatime` formatına salınaraq il və ay sütunları yaradıldı
*   Məlumat tipləri yoxlanıldı

2.Exporatory Data Analysis(EDA)
*   İllər üzrə Netflix-ə əlavə olunan kontentlərin artım trendi analiz edildi
*   Kontent növlərinin nisbəti hesablandı

3.KPI və Pivot Table Analysis
*   Illər və aylar üzrə kontent artımını görmək üçün `Pivot Table` quruldu

4.Data Vizualization
*   `Seaborn` və `Matplotlib` vasitəsilə Barplot,Lineplot,Pie Chart və Heatmap qrafikləri quruldu

#Əsas nəticələr
*   **Kontent Paylanması:**Netflix kitabxanasının təxminən **69.6%-i Filmlər**,**30.4%-i isə TV Şoular(Seriallar)** təşkil edir.
*   **Lider Ölkələr:** Yayımlanan kontent sayına görə **ABŞ(USA)** açıq fərqlə birincidir,onu **Hindistan(İndia)** və **Böyük Britaniya(UK)** izləyir
*   **Tarix Trendi:** 2016-cı ildən etibarən Netflix platformasınaəlavə olunan film və seriallarınsayında kəskin artım müşahidə olunub,ən pik nöqtə 2018-2020-ci illərdə olmuşdur.
